**Make sure to not have any spaces**

```python
df_filtered.loc[:, "design_models"] = df_filtered["design_models"].str.replace(" ", "_")
df_filtered
```

**Convert to a FASTA File**

```python
def df_to_fasta(df, name_col="id", seq_col="sequence", output="sequences.fasta"):
    with open(output, "w") as f:
        for _, row in df.iterrows():
            name = str(row[name_col]).strip().replace("-", "_")
            seq = str(row[seq_col]).strip()
            if pd.notna(name) and pd.notna(seq):
                f.write(f">{name}\n{seq}\n")
```

```python
df_to_fasta(df2, name_col='id', seq_col='sequence', output='contest_2_round_2_sequences.fasta')
```

**Explode just like in the large data set**

```python
import json

df2["evaluations"] = df2["evaluations"].apply(json.loads)

# print(type(alldata.loc[0, "evaluations"]))

df_metrics = df2.explode("evaluations", ignore_index=True)
df_metrics = df_metrics.join(pd.json_normalize(df_metrics.pop("evaluations")))
df_metrics
```

**Drop irrelevant columns**

```python
def drop_columns(df, columns):
    """
    Drop the specified columns from a pandas dataframe.
    """
    return df.drop(columns=columns, axis=1, errors='ignore')

columns_to_drop = ['value.url']
df_metrics = drop_columns(df_metrics, columns_to_drop)
df_metrics
```

**Make a pivot table and export to CSV**

```python
df_metrics["designMethod"] = df_metrics["designMethod"].fillna("Unknown")
subset_pivot = df_metrics.pivot_table(index=["name", "id", "designMethod", "sequence"], columns="metric", values="value", aggfunc="first").reset_index()
subset_pivot.columns.name = None
subset_pivot
# subset_pivot.to_csv('contest_2_round_2_metrics_summary.csv', index=False)
```
