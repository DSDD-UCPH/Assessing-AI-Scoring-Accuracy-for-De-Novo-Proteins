```python
import pandas as pd
import os
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import json
```

**First part is to extract the data for each run and create a compiled CSV file**

```python
# Path to folder with all the predictions.
root = "/home/postyr/Assessing-AI-Scoring-Accuracy-for-De-Novo-Proteins/results/Boltz-2/Boltz-2_run1"

records = []

for complex_folder in os.listdir(root):
    folder_path = os.path.join(root, complex_folder)
    if not os.path.isdir(folder_path):
        continue

# Find the summary json files with the iptm values.

    for fname in os.listdir(folder_path):
        if fname.endswith("model_0.json"):
            json_path = os.path.join(folder_path, fname)

# name is the folder name
            name = complex_folder

# Load JSON
            with open(json_path, "r") as f:
                data = json.load(f)

            records.append({
                "name": name,
                "confidence_score": data.get("confidence_score"),
                "iptm": data.get("iptm"),
                "ptm": data.get("ptm"),
                "ligand_iptm": data.get("ligand_iptm"),
                "protein_iptm": data.get("protein_iptm"),
                "complex_plddt": data.get("complex_plddt"),
                "complex_iplddt": data.get("complex_iplddt"),
                "complex_pde": data.get("complex_pde"),
                "complex_ipde": data.get("complex_ipde"),
                "pair_chains_iptm": data.get("pair_chains_iptm"),
                "chains_ptm": data.get("chains_ptm"),
            })

df_json = pd.DataFrame(records)
print(df_json.head())
print("Number of complexes found:", len(df_json))
df_json.to_csv("/home/postyr/Assessing-AI-Scoring-Accuracy-for-De-Novo-Proteins/results/Boltz-2/metrics_merged.csv", index=False)
```
